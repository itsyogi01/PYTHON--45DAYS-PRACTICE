```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# from a pandas dataframe
batsman = pd.read_csv("/content/sharma-kohli.csv")
batsman
plt.plot(batsman["index"],batsman['V Kohli'])
plt.plot(batsman['index'],batsman['RG Sharma'])
```

```python
# label title
plt.plot(batsman["index"],batsman['V Kohli'])
plt.plot(batsman['index'],batsman['RG Sharma'])

plt.title("Rohit Sharma Vs Virat Kholi Carrier Compersion")
plt.xlabel('Season')
plt.ylabel('Runs Scored')
```

```python
# color(hex) and line(width and style) and marker(size)
plt.plot(batsman["index"],batsman['V Kohli'],color='green')
plt.plot(batsman['index'],batsman['RG Sharma'],color='black')

plt.title("Rohit Sharma Vs Virat Kholi Carrier Compersion")
plt.xlabel('Season')
plt.ylabel('Runs Scored')
```

```python
plt.plot(batsman["index"],batsman['V Kohli'],color='#F21C02')
plt.plot(batsman['index'],batsman['RG Sharma'],color='#568551')

plt.title("Rohit Sharma Vs Virat Kholi Carrier Compersion")
plt.xlabel('Season')
plt.ylabel('Runs Scored')
```

```python
# from matplotlib.lines import lineStyles
plt.plot(batsman["index"],batsman['V Kohli'],linestyle='dashed')
plt.plot(batsman['index'],batsman['RG Sharma'],linestyle='dashed')

plt.title("Rohit Sharma Vs Virat Kholi Carrier Compersion")
plt.xlabel('Season')
plt.ylabel('Runs Scored')
```

```python
plt.plot(batsman["index"],batsman['V Kohli'],linestyle='dotted')
plt.plot(batsman['index'],batsman['RG Sharma'],linestyle='dotted')

plt.title("Rohit Sharma Vs Virat Kholi Carrier Compersion")
plt.xlabel('Season')
plt.ylabel('Runs Scored')
```

```python
plt.plot(batsman["index"],batsman['V Kohli'],linestyle='dashdot')
plt.plot(batsman['index'],batsman['RG Sharma'],linestyle='dashdot')

plt.title("Rohit Sharma Vs Virat Kholi Carrier Compersion")
plt.xlabel('Season')
plt.ylabel('Runs Scored')
```

```python
plt.plot(batsman["index"],batsman['V Kohli'],linestyle='solid',linewidth=3)
plt.plot(batsman['index'],batsman['RG Sharma'],linestyle='dotted')

plt.title("Rohit Sharma Vs Virat Kholi Carrier Compersion")
plt.xlabel('Season')
plt.ylabel('Runs Scored')
```

```python
plt.plot(batsman["index"],batsman['V Kohli'],linestyle='dotted',marker ="D")
plt.plot(batsman['index'],batsman['RG Sharma'],linestyle='dotted',marker ="+")

plt.title("Rohit Sharma Vs Virat Kholi Carrier Compersion")
plt.xlabel('Season')
plt.ylabel('Runs Scored')
```

```python
plt.plot(batsman["index"],batsman['V Kohli'],linestyle='dotted',marker ="D",markersize = 10)
plt.plot(batsman['index'],batsman['RG Sharma'],linestyle='dotted',marker ="+")

plt.title("Rohit Sharma Vs Virat Kholi Carrier Compersion")
plt.xlabel('Season')
plt.ylabel('Runs Scored')
```

```python
plt.plot(batsman["index"],batsman['V Kohli'],linestyle='dotted',marker ="D",markersize = 10,label='virat')
plt.plot(batsman['index'],batsman['RG Sharma'],linestyle='dotted',marker ="+",label='rohit')

plt.title("Rohit Sharma Vs Virat Kholi Carrier Compersion")
plt.xlabel('Season')
plt.ylabel('Runs Scored')


plt.legend()
# plt.legend(loc = "best")
# plt.legend(loc = "upper right")
```

```python
# =======================================bar chart==============================
# bivarite and univarite
# numerical and categorical
# use case-Aggregate analysis of group
```

```python
# simple bar chart
children = [10,20,30,40,50]
colors = ['red','blue','green','yellow','pink']
plt.bar(colors,children,color='pink')
plt.title("child like color")

plt.ylabel('children')
plt.xlabel('colors')
```

```python
# vertical bar chart and horizontal bar chart
plt.barh(colors,children,color='pink')
```
