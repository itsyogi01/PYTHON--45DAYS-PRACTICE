```python
# -------------DUCK TYPING-----------------
class Pigeon:
    def fly(self):
        print("Pigeon flying")

class Airplane:
    def fly(self):
        print("Airplane flying")

def start_flying(obj):
    obj.fly()

start_flying(Pigeon())
start_flying(Airplane())
```

```python
# -----------------COMPOSITION IN PYTHON-----------------

class Engine:
    def start(self):
        print("Engine started")

class Car:
    def __init__(self):
        self.engine = Engine()   # Car Ke Andhar Engine object
    def drive(self):
        self.engine.start()
        print("Car is moving")

c = Car()
c.drive()
```
