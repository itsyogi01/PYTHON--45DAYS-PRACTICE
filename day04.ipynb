```python
# using with statement
# Auto close file
with open("example.txt","w+") as f:
  f.write("this file is creating for the implemention of woth statement")
  f.seek(0)
  data = f.read()
  print(data)
```

```python
# copy contant from one file to another file
with open("newfile.txt","r") as src:
  content = src.read()

with open("newfile3.txt","w+") as dest:
  dest.write(content)
  dest.seek(0)
  data = dest.read()
  print(data)
```

```python
# EXAMPLE 1
#  Student records system
with open("students.txt","w") as s:
    s.write("Nikita, 20\n")
    s.write("Radha, 20\n")
    s.write("Rupesh, 20\n")

with open("students.txt","r") as f:
  for line in f:
    name, age = line.strip().split(",")
    print(f"NAME: {name} , Age: {age}")
```

```python
# Example 2
# --------------word counter in program----------------
with open("story.txt","w") as f:
   f.write("Python is easy to learn.\nPython is powerful. \n")
# count words
with open("story.txt","r") as f:
  text = f.read()
  words = text.split()
  print("total words:",len(words))
```

```python
# Example 3
# ------------------searhing in a file------------------
with open("notes.txt","w") as f:
  f.write("Every day is not good somthing is good in everyday.\nshow enjoy your life ")
search_word = "is"

with open("notes.txt","r") as f:
    for line in f:
      if search_word in line:
        print("found:",line.strip())
```
