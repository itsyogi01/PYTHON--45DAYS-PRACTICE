```python
# -------------ENCAPSULATION IN PYTHON------------
class BankAccount:
  def __init__(self,balance):
    self.__balance = balance
  def deposite(self,ammount):
      self.__balance += ammount
  def withdraw(self,ammount):
       self.__balance -= ammount
  def get_balance(self):
      return self.__balance
A1 = BankAccount(20000)
A1.deposite(2000)
A1.withdraw(200)
print(A1.get_balance())
```

```python
# ---------------inheritance--------------------
# single inheritance
class Parent:
  def show(self):
    print("this is a parent class ")
class Child(Parent):
    def display(self):
      print("this is child class")

c = Child()
c.show()
c.display()
```

```python
# -------------------multiple inheritance------------
class A:
     car = "BMW"
     plot = "Jaipur"
     Property = "23 cr"
     def grandfather(self):
        print("total property = " + A.Property)
class B(A):
    bike = "hero xtrem"
    def father(self,bike):
      print(bike)

class C(B):
    scooty = "Ather"
    def child(self,scooty):
      print(scooty)

ob = C()
print(ob.scooty,ob.bike,ob.Property,ob.plot,ob.car)
ob.grandfather()
ob.father("hero")
ob.child("activa")
```

```python
# ----------------polymorphism--------------------
# method overloading
class math:
   def add(self ,a,b=3,c=4):
       return a+b+c
c = math()
print(c.add(10,20))
print(c.add(10))
print(c.add(10,20,30))





# method overiding
class Animal:
  def sound(self):
    print("Animal sound")
class Dog(Animal):
  def sound(self):
    print("Bark")
d = Dog()
d.sound()
```

```python
# --------------------Abstraction in python-----------------
from abc import ABC,abstractmethod
class Shape(ABC):
   @abstractmethod
   def area(self):
     pass
class Circle(Shape):
  def __init__(self,radius):
    self.radius = radius
  def area(self):
      return 3.14*self.radius*self.radius

c = Circle(4)
print("Area", c.area())
```

```python
# ----------------static methos and class method-------------
class Demo:
  count = 0
  def __init__(self):
    Demo.count +=1

  @staticmethod
  def greet():
    print("Hello!")

  @classmethod
  def show_count(cls):
    print("object created:",cls.count)

d1 = Demo()
d2 = Demo()
Demo.greet()
Demo.show_count()
```

```python
# --------------------Operater Overloading----------------------
class Point:
    def __init__(self,x,y):
       self.x,self.y = x,y
    def __add__(self,other):
     return Point(self.x+other.x,self.y+other.y)

p1 = Point(10,20)
p2 = Point(30,40)
p3 = p1+p2
print(p3.x , p3.y)
```

```python
# -----------------------. Data Hiding & Property Decorator------------------
class Employee:
  def __init__(self,salary):
    self.__salary = salary

  @property
  def salary(self):
    return self.__salary

  @salary.setter
  def salary(self,value):
    if value > 0:
      self.__salary = value
e = Employee(100000)
print(e.salary)
e.salary = 7000
print(e.salary)
```

```python
# --------------------------Getter----------------------
class Student:
    def __init__(self, name):
        self.__name = name   # private variable

    @property
    def name(self):         # getter
        return self.__name

s = Student("Rupesh")
print(s.name)   # Output: Rupesh
```

```python
class Student:
    def __init__(self, name):
        self.__name = name

    @property
    def name(self):         # getter
        return self.__name

    @name.setter
    def name(self, value):  # setter
        if value != "":
            self.__name = value

s = Student("Rupesh")
print(s.name)   # Rupesh
s.name = "Rahul"
print(s.name)   # Rahul
```

```python
class Employee:
    def __init__(self, salary):
        self.__salary = salary

    # Getter
    @property
    def salary(self):
        return self.__salary

    # Setter with validation
    @salary.setter
    def salary(self, value):
        if value <= 0:
            raise ValueError("Salary must be positive!")
        self.__salary = value

e = Employee(5000)
print(e.salary)      # 5000

e.salary = 7000      # update via setter
print(e.salary)      # 7000

# e.salary = -1000   # ❌ Error: Salary must be positive!
```
